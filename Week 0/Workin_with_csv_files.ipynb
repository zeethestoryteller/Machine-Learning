# Opening a CSV file from url

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

# Sep Parameter

we use this because some times when we use tab separated values then "read.csv" doen't work so we change sep parameter from comma to tab.

```python
pd.read_csv("data.csv", sep="\t")
```

if coloumn name are not showing then we pass a list and give coloumns a name

```python
pd.read_csv("data.csv", sep="\t",names=["c1","c2",...])
```

# Index_col parameter

```python
pd.read_csv("data.csv",index_col=["col_name"])
```

# Header Parameter

When our colomn name used as a row

```python
pd.read_csv("data.csv",header=1)
```

# use_cols parameter

when we want to work only desired columns

```python
pd.read_csv("data.csv",usecols=["col1","col2"])
```

Squeeze Parameters: When we need only one column

```python
pd.read_csv("data.csv",usecols=["col1"],squeeze=True)
```

If want to skip some rows from data then we use

#Skiprows/nrows Parameter

```python
pd.read_csv("data.csv",skiprows=[1,3])
```

It will skip row 1,3 from the data

```python
pd.read_csv("data.csv",nrows=100)
```

it will show n rows from top of the data

# Skip Bad Lines

When, in a row values are more then columns. It skip that lines that have extra values

```python
pd.read_csv("data.csv",error_bad_lines=False)
```

# dtypes Parameter

In pandas jab read_csv function data ko dekhta hai tab vo analyse kar leta hai ki columns me kis type ki values hai but we want to overcome this behavior then we use this paarmeter

```python
pd.read_csv("data.csv",dtype={"col1":"changed to"})
```

# Handling Dates

read_csv read dates as stings.So we use this function to format date string to date.

```python
pd.read_csv("data.csv",parse_dates=["col1"])
```
